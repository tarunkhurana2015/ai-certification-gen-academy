# Sample scenario notebooks — index

The learning notebook has been split into focused files under [`server/sample_scenarios/`](sample_scenarios/README.md).

Open the [learning index and prerequisites](sample_scenarios/README.md), then choose the lesson you want. All original lesson content and saved outputs are retained in those files. This entry-point notebook intentionally contains no executable cells.

Each executable lesson includes its own source-only Python setup. Use the project's `.venv` kernel and run that lesson top to bottom. App navigation state persists separately from Python kernel state; see the per-lesson device-state prerequisites.
